```sage
q = 256
K = GF(q, name='ξ')
R = PolynomialRing(K, name='w')
R_ = PolynomialRing(GF(2), name='x')
w = R.gen()
ξ = K.gen()
x = R_.gen()

```

```sage
n = q - 1
f = (w^n - 1)/(w-1) 

polys = [(j,(ξ^j).minpoly()) for j in range(1,n)]
grouped = dict()
for (j,p) in polys:
    if p in grouped:
        grouped[p].add(j)
    else:
        grouped[p] = set([j])


import random as rn

def gg(l : int):
    assert l % 2 == 0, f"argumento deve ser par"
    g_ = set([]) ; pp = list(grouped) ; r_ = set()
    while l > 0:
        p = rn.choice(pp) ; s = p.degree()  
        if l >= s and p not in g_:
            g_.add(p) ; r_.update(grouped[p]); l = l - s 
    return prod(g_), set([ξ^j for j in r_])
        
```

```sage
k = 223
g = gg(n - k)
```

```sage
m = R_.random_element(k)

def rsc(m):
    m_ = m * x^(n-k)
    return m_ + (m_ % g)

def synd(m_):
    return m_ % g
```
